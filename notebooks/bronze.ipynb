```python
from pyspark.sql.functions import current_timestamp, lit

df = spark.read.option("header", True).option("inferSchema", True).csv("Files/revenue_transactions_3yr_100k.csv")
df = df.withColumn("_ingested_at", current_timestamp()).withColumn("_source_file", lit("revenue_transactions_3yr_100k.csv"))
df.write.format("delta").mode("overwrite").saveAsTable("bronze_revenue_transactions")
```
